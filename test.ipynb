```python
import requests
import json
import os
import time
import logging
from pathlib import Path
import pandas as pd
import os
from dotenv import load_dotenv

load_dotenv()
keycloak_url = "https://keycloak.tp.entsoe.eu/realms/tp/protocol/openid-connect/token"
fms_url = "https://fms.tp.entsoe.eu/listFolder"
username = os.getenv("ENTSOE_USER")
password = os.getenv("ENTSOE_PWD")
client_id = "tp-fms-public"

KEYCLOAK_URL = "https://keycloak.tp.entsoe.eu/realms/tp/protocol/openid-connect/token"
LIST_URL = "https://fms.tp.entsoe.eu/listFolder"
DOWNLOAD_URL = "https://fms.tp.entsoe.eu/downloadFileContent"

racine_locale = Path(r"C:\Users\PC\Documents\Projets\TFE\Datas\Electricite")
racine_locale.mkdir(parents=True, exist_ok=True)

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(message)s",
    handlers=[
        logging.FileHandler(racine_locale / "ingestion_electricite.log", encoding="utf-8"),
        logging.StreamHandler(),
    ],
)
log = logging.getLogger("entsoe_elec")

# ============================================================
# Gestion du token : rafraîchi automatiquement avant expiration
# ============================================================
TOKEN_DUREE_MAX = 8 * 60  # rafraîchit après 8 min, avant l'expiration à 10 min

_token_cache = {"access_token": None, "obtenu_a": 0}


def obtenir_token():
    payload_auth = {
        "client_id": client_id,
        "grant_type": "password",
        "username": username,
        "password": password,
    }
    headers_auth = {'Content-Type': 'application/x-www-form-urlencoded'}
    resp_auth = requests.post(KEYCLOAK_URL, headers=headers_auth, data=payload_auth)
    resp_auth.raise_for_status()
    return resp_auth.json()["access_token"]


def get_headers():
    maintenant = time.time()
    if _token_cache["access_token"] is None or (maintenant - _token_cache["obtenu_a"]) > TOKEN_DUREE_MAX:
        _token_cache["access_token"] = obtenir_token()
        _token_cache["obtenu_a"] = maintenant
        log.info("Token rafraîchi")
    return {
        'Authorization': f'Bearer {_token_cache["access_token"]}',
        'Content-Type': 'application/json',
        'Accept': '*/*',
    }


# ============================================================
# Téléchargement
# ============================================================
metriques = []

# --- 1) liste tous les dossiers à la racine ---
payload = json.dumps({"path": "/TP_export/", "pageInfo": {"pageIndex": 0, "pageSize": 5000}})
resp = requests.post(LIST_URL, headers=get_headers(), data=payload)
resp.raise_for_status()
dossiers = resp.json()["contentItemList"]
log.info(f"{len(dossiers)} dossiers trouvés sous /TP_export/")

# --- 2) pour chaque dossier, liste et télécharge les CSV ---
for dossier in dossiers:
    nom_dossier = dossier["name"]
    chemin_local_dossier = racine_locale / nom_dossier
    chemin_local_dossier.mkdir(parents=True, exist_ok=True)

    payload = json.dumps({
        "path": f"/TP_export/{nom_dossier}/",   # listFolder garde le slash final
        "pageInfo": {"pageIndex": 0, "pageSize": 5000},
    })
    resp = requests.post(LIST_URL, headers=get_headers(), data=payload)
    resp.raise_for_status()
    fichiers = resp.json()["contentItemList"]

    for fichier in fichiers:
        nom_fichier = fichier["name"]
        chemin_local_fichier = chemin_local_dossier / nom_fichier
        debut = time.perf_counter()

        try:
            payload_dl = json.dumps({
                "folder": f"/TP_export/{nom_dossier}",   # PAS de slash final ici
                "filename": nom_fichier,
                "topLevelFolder": "TP_export",
                "downloadAsZip": False,
            })
            resp_dl = requests.post(DOWNLOAD_URL, headers=get_headers(), data=payload_dl)
            resp_dl.raise_for_status()
            with open(chemin_local_fichier, "wb") as f:
                f.write(resp_dl.content)
        except requests.exceptions.RequestException as e:
            log.error(f"{nom_dossier} | {nom_fichier} | échec : {e}")
            metriques.append({"dossier": nom_dossier, "fichier": nom_fichier, "taille_ko": 0, "duree_s": 0})
            continue

        duree = time.perf_counter() - debut
        taille_ko = os.path.getsize(chemin_local_fichier) / 1024
        log.info(f"{nom_dossier} | {nom_fichier} | {taille_ko:.0f} Ko | {duree:.2f} s")
        metriques.append({"dossier": nom_dossier, "fichier": nom_fichier,
                           "taille_ko": taille_ko, "duree_s": duree})

pd.DataFrame(metriques).to_csv(racine_locale / "metriques_electricite_ingestion.csv", index=False)
log.info(f"Terminé. {len(metriques)} fichiers traités.")
```
