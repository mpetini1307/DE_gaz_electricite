```python
import pandas as pd
import requests
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
api_key = os.getenv("COUNTRY_API_KEY")
```

```python
url = "https://api.restcountries.com/countries/v5"
params = {"limit": 5, "offset": 0}
countries = []

while True:
    contenu = requests.get(url, headers={"Authorization": api_key}, params=params).json()

    for p in contenu["data"]["objects"]:
        countries.append({
            "code": p["codes"]["alpha_2"],
            "pays": p["names"]["common"],
            "latitude": p["coordinates"]["lat"],
            "longitude": p["coordinates"]["lng"],
        })

    if not contenu["data"]["meta"]["more"]:
        break
    params["offset"] += 5

df_countries = pd.DataFrame(countries)
df_countries.to_csv(rf'C:\Users\PC\Documents\Projets\TFE\Datas\Pays\pays_infos.csv', index=False)
```
