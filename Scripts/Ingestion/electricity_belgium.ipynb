```python
import pandas as pd
import os
from os import listdir
from os.path import join
import time
```

```python
folders = ['ActualTotalLoad_6.1.A_r3',
          'AggregatedGenerationPerType_16.1.B_C_r3',
          'PhysicalFlows_12.1.G_r3',
          'InstalledGenerationCapacityAggregated_14.1.A_r3'
          ]
```

```python
metriques = []
for folder in folders:
    path = rf'C:\Users\PC\Documents\Projets\TFE\Datas\Electricite\{folder}'
    countryCode = 'BE'
    writing_path = rf'C:\Users\PC\Documents\Projets\TFE\Datas\Electricite\Belgium'
    if not os.path.exists(join(writing_path,folder)): os.makedirs(join(writing_path,folder))
    files = [f for f in listdir(path)]
    for file in files:
        path_file = rf'C:\Users\PC\Documents\Projets\TFE\Datas\Electricite\Belgium\{folder}\{file}_BE.csv'
        debut = time.perf_counter()

        df = pd.read_csv(join(path,file), sep='\t')
        if folder == 'PhysicalFlows_12.1.G_r3' : 
            df = df[df['InAreaMapCode'] == countryCode]
        else : 
            df = df[df['AreaMapCode'] == countryCode]
        lines = df.shape[0]
        df.to_csv(path_file)

        duree = time.perf_counter() - debut
        taille_ko = os.path.getsize(path_file) / 1024   
        print(folder,'\n',file,'\n',taille_ko,'\n',duree,'\n',lines,'\n')     
        metriques.append({"dossier":folder, "fichier": file,
                        "taille_ko": taille_ko, "duree_s": duree, "lignes":lines})


pd.DataFrame(metriques).to_csv(rf"C:\Users\PC\Documents\Projets\TFE\Datas\Electricite\Belgium\metriques_electricite_BE.csv", index=False)
```
